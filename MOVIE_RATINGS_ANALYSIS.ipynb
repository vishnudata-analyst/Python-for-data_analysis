# **Project: Movie Ratings Analysis Using NumPy, Pandas, and Matplotlib**

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

**Importing file for analysis **

```python
from google.colab import files
uploaded = files.upload()  # ye ek "Choose File" button dega

import pandas as pd
df = pd.read_csv('movie_ratings_analysis_dataset.csv')  # jo naam upload hua wahi use karo
df.head()
```

**Using info() to check the row,columns and datatypes**

```python
df.info()
```

**We will use the isnull()to check the null values in the dataset**

```python
df.isnull().sum()
```

**We use describe to check the statistics report**

```python
df["Rating"].describe()
```

**We will use fillna to fill the null value**

```python
df["Rating"].fillna(df["Rating"].mean(),inplace=True)
```

**Checking wheather the action done in right way or not**

```python
df["Rating"].isnull().sum()
```

```python
df["Budget"].describe()
```

```python
df["Budget"].fillna(df["Budget"].median(),inplace=True)
```

```python
df["Budget"].isnull().sum()
```

```python
df["Revenue"].describe()
```

```python
df["Revenue"].fillna(df["Revenue"].median(),inplace=True)
```

```python
df["Revenue"].isnull().sum()
```

```python
df["Director"].fillna("Unknown",inplace=True)
```

```python
df["Director"].isnull().sum()
```

**Now we will find out the average rating of the movies **

```python
np.mean(df["Rating"])
```

```python
np.median(df["Rating"])
```

```python
np.std(df["Rating"])
```

**Now will use pandas library to perform the group analysis**

```python
df.groupby("Genre")["Rating"].mean()
```

```python
df["Genre"].value_counts()
```

```python
df.groupby("Director")["Rating"].mean()
```

```python
df.groupby("Year")["Rating"].mean()
```

**Now we will use matplotlib library to visualize our findings through visual like charts**

```python
genre_rating=df.groupby("Genre")["Rating"].mean()
genre_rating
```

# **BAR CHART**

```python
plt.bar(genre_rating.index,genre_rating.values)
plt.title("Average_rating_by_Genre")
plt.xlabel("Genre")
plt.ylabel("Average_Rating")
plt.xticks(rotation=45)
plt.show()
```

```python
year_rating=df.groupby("Year")["Rating"].mean()
year_rating
```

# **LINE CHART**

```python
plt.plot(year_rating.index,year_rating.values,marker="o")
plt.title("Average_rating_by_year")
plt.xlabel("Year")
plt.ylabel("Average_rating")
plt.grid()
plt.show()
```

# **SCATTER PLOT**

```python
plt.scatter(df["Budget"],df['Rating'])
```

```python
df["Budget"].corr(df["Rating"])
```

```python
plt.scatter(df["Revenue"],df["Rating"])
```

```python
top_movies=df.nlargest(10,"Rating")
top_movies[["Title","Rating"]]
```

# **HORIZOTAL BAR**

```python
plt.barh(top_movies["Title"],top_movies["Rating"])
plt.title("top_10 rated movies")
plt.xlabel("Rating")
plt.ylabel("Movie")
plt.gca().invert_yaxis()
plt.show()
```
